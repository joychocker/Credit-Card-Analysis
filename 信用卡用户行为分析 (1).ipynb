### *明确列名含义
Attrition_Flag（客户流失标志）：标明客户是否已经流失

Dependent_count（抚养子女人数）：客户需要抚养的子女数量，影响其家庭财务负担。

Education_Level（教育程度）：最高教育水平

Marital_Status（婚姻状况）

Card_Category（信用卡类别）：持有的信用卡等级，从低到高为：Blue(蓝卡)->Silver(银卡)->Gold(金卡)->Platinum(白金卡)

Months_on_book（在册月数）：客户成为银行客户的总月数。时间越长，忠诚度可能越高。

Total_Relationship_Count（总关系数量）：客户在银行持有的产品总数，反映客户与银行的绑定程度。

Months_Inactive_12_mon（过去12个月不活跃月数）： 在过去12个月中，客户没有任何交易活动的月数，与流失风险成正比。

Contacts_Count_12_mon（过去12个月联系次数）：银行客服主动联系该客户的次数。联系多可能是因为客户有投诉、或银行在进行挽留。

Credit_Limit（信用额度）：银行授予该信用卡的最高可透支金额。比如额度是$10,000。

Total_Revolving_Bal（总循环余额）：客户逾期欠款余额，银行收取余额利息。

Avg_Open_To_Buy（平均可用额度）：过去12个月内，“信用额度-循环余额”的均值。

Total_Trans_Amt（总交易金额）：过去12个月内，客户用该信用卡消费的总金额。

Total_Trans_Ct（总交易次数）：过去12个月内，客户用该信用卡消费的总次数。

Total_Amt_Chng_Q4_Q1（Q4与Q1交易金额变化率）

Total_Ct_Chng_Q4_Q1（Q4与Q1交易次数变化率）

Avg_Utilization_Ratio（平均使用率）：循环余额/信用额度，使用率过低（如<10%）可能是“沉睡客户”；使用率过高（如>80%）可能表示客户财务状况紧张，风险较高。

**Step5 特征工程(Feature Engineering)**

是数据分析和机器学习中非常重要的一步，它是指利用领域知识从原始数据中创建新的特征，以提高模型的性能或更好地理解数据。

```python
# “客户年龄段”
bins = [20, 30, 40, 50, 60, 100]
labels = ['20-30岁', '31-40岁', '41-50岁', '51-60岁', '60岁以上']
df['Age_Group'] = pd.cut(df['Customer_Age'], bins=bins, labels=labels)

# “客户活跃度分级”
def activity_level(inactive_months):
    if inactive_months <= 1:
        return '高活跃度'
    elif inactive_months <= 2:
        return '中活跃度'
    else:
        return '低活跃度'
df['Activity_Level'] = df['Months_Inactive_12_mon'].apply(activity_level)

# “信用卡使用强度”
def usage_intensity(ratio):
    if ratio == 0:
        return '未使用'
    elif ratio <= 0.3:
        return '轻度使用'
    elif ratio <= 0.7:
        return '中度使用'
    else:
        return '重度使用'
df['Usage_Intensity'] = df['Avg_Utilization_Ratio'].apply(usage_intensity)

print("✅ 特征工程完成！")
print("\n✨ 新增特征预览：")
display(df[['Customer_Age', 'Age_Group', 'Months_Inactive_12_mon', 'Activity_Level', 'Avg_Utilization_Ratio', 'Usage_Intensity']].head(10))
```

```python
# 研究不同类别下流失客户比例
# 选取:性别/教育程度/婚姻状况/收入水平/卡片类别
categorical_cols = ['Gender', 'Education_Level', 'Marital_Status', 'Income_Category', 'Card_Category']

print('🔵初步分类分析流失客户是否存在性别/教育程度/婚姻状况/收入水平/卡片类别的显著特征')

for col in categorical_cols:
    # 计算每个类别下流失和现有客户的数量
    attrition_counts = df.groupby([col, 'Attrition_Flag']).size().unstack(fill_value=0)
    # 计算流失率
    attrition_rates = attrition_counts.apply(lambda x: x / x.sum(), axis=1)

    # 将比例格式化为百分比字符串
    attrition_rates_formatted = attrition_rates.applymap(lambda x: f'{x:.1%}')

    attrition_rates_with_text = attrition_rates.copy()
    for flag in attrition_rates.columns:
        attrition_rates_with_text[f'{flag} (%)'] = attrition_rates_formatted[flag]


    # 创建堆叠柱状图
    fig = px.bar(attrition_rates_with_text, # Use the DataFrame with formatted text columns
                 title=f'按 {col} 分类的客户流失率',
                 labels={'value': '比例', col: col, 'variable': '客户状态'},
                 color_discrete_map={'Existing Customer': 'blue', 'Attrited Customer': 'red'},
                 hover_data=[f'{flag} (%)' for flag in attrition_rates.columns] # Explicitly include formatted text columns in hover_data
                )

    fig.update_layout(barmode='stack')
    fig.show()
```

### Step7  细化&量化客户分群

为更客观和系统地进行客户分群，选择K-Means聚类。(K-Means聚类是一种常用的无监督学习算法，其根据数据的相似性将客户分成K个不同的群体。)

1.   使用手肘法确定最佳聚类数量K=4
2.   将用户分为4类群



```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

features_for_clustering = ['Credit_Limit', 'Total_Revolving_Bal', 'Total_Trans_Amt', 'Total_Trans_Ct', 'Months_on_book']
df_clustering = df[features_for_clustering].copy()

scaler = StandardScaler()
df_scaled = scaler.fit_transform(df_clustering)

# 使用“手肘法”确定最佳的聚类数量 K
sse = [] # Sum of squared errors,误差平方和判断聚合好坏
k_range = range(1, 11)
for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10) # Add n_init to suppress warning
    kmeans.fit(df_scaled)
    sse.append(kmeans.inertia_)

# 绘制手肘图
fig = px.line(x=list(k_range), y=sse, markers=True,
              title='使用手肘法确定最佳聚类数量 K',
              labels={'x': '聚类数量 (K)', 'y': 'SSE'})
fig.update_layout(xaxis = dict(tickmode = 'linear', tick0 = 1, dtick = 1),
                  title_x=0.5) # Add title_x=0.5 to center the title
fig.show()
```

```python
# 使用 K=4 执行 K-Means 聚类
optimal_k = 4

# 执行 K-Means 聚类
kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
df['Cluster'] = kmeans.fit_predict(df_scaled)

print(f"✅ K-Means 聚类完成，已将聚类结果添加到 'Cluster' 列。")

# 分析每个聚类群体的特征
print("\n✨ 各聚类群体的特征分析：")
cluster_analysis = df.groupby('Cluster')[features_for_clustering + ['Attrition_Flag']].agg({
    'Credit_Limit': 'mean',
    'Total_Revolving_Bal': 'mean',
    'Total_Trans_Amt': 'mean',
    'Total_Trans_Ct': 'mean',
    'Months_on_book': 'mean',
    'Attrition_Flag': lambda x: (x == 'Attrited Customer').sum() / len(x) * 100 # 计算流失率并转换为百分比
}).reset_index()

# 重命名流失率列
cluster_analysis = cluster_analysis.rename(columns={'Attrition_Flag': 'Attrition_Rate (%)'})

# 格式化输出，保留两位小数
cluster_analysis = cluster_analysis.round(2)

display(cluster_analysis)

# 可视化聚类结果 (例如，在信用额度 vs 循环余额散点图上按聚类着色)
fig_cluster_scatter = px.scatter(df, x='Credit_Limit', y='Total_Revolving_Bal', color='Cluster',
                                 hover_data=features_for_clustering + ['Attrition_Flag'],
                                 title='💸 信用额度 vs. 循环余额 (按聚类着色)',
                                 labels={'Credit_Limit': '信用额度', 'Total_Revolving_Bal': '循环余额'},
                                 color_continuous_scale=px.colors.qualitative.Plotly) # 使用Plotly内置色板

fig_cluster_scatter.show()

# 可视化总交易金额 vs 总交易笔数 按聚类着色
fig_cluster_trans = px.scatter(df, x='Total_Trans_Amt', y='Total_Trans_Ct', color='Cluster',
                               hover_data=features_for_clustering + ['Attrition_Flag'],
                               title='📈 总交易金额 vs. 总交易笔数 (按聚类着色)',
                               labels={'Total_Trans_Amt': '总交易金额', 'Total_Trans_Ct': '总交易笔数'},
                               color_continuous_scale=px.colors.qualitative.Plotly) # 使用Plotly内置色板

fig_cluster_trans.show()
```

*   **Cluster 2 (高信用额度潜在客户):** 信用额度非常高，但交易活跃度中等。有较高潜力，需激活。
*   **Cluster 0 (中等活跃度客户):** 交易活跃度中等，循环余额相对较高。是银行的基础客户群体。
