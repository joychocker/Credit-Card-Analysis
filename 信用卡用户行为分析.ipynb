**Step4 数据预处理**（清洗+准备工作）

```python
## 数据预处理 清洗+准备工作
# 1. 创建原始数据的一个副本，以便在副本上操作，保留原始数据
df = df_real.copy()

print("✅ 创建数据副本完成")

# 2. 检查并处理明显的异常值 (例如：年龄为0或超过100)
print(f"年龄范围: {df['Customer_Age'].min()} 到 {df['Customer_Age'].max()}")

# 3. 删除与分析无关的列
# CLIENTNUM是客户ID，没有分析价值。
# 两个Naive_Bayes_Classifier列是模型预测的中间结果，我们不做预测，所以也删除。
columns_to_drop = ['CLIENTNUM', 'Naive_Bayes_Classifier_Attrition_Flag_Card_Category_Contacts_Count_12_mon_Dependent_count_Education_Level_Months_Inactive_12_mon_1', 'Naive_Bayes_Classifier_Attrition_Flag_Card_Category_Contacts_Count_12_mon_Dependent_count_Education_Level_Months_Inactive_12_mon_2']
df = df.drop(columns=columns_to_drop)

print(f"✅ 无关列已删除。新的数据形状: {df.shape}")

# 4. 查看处理后的数据
print("\n处理后的数据前5行：")
display(df.head())# .head()函数默认5行,对应倒数函数为.tail()
display(df.tail(3))
```

**Step5 基于原始数据 进行特征分析(Feature Engineering)**
特征工程是数据分析和机器学习中非常重要的一步，它是指利用领域知识从原始数据中创建新的特征，以提高模型的性能或更好地理解数据。

```python
print("🎯 开始特征工程...")

# 1. 创建“客户年龄段”
bins = [20, 30, 40, 50, 60, 100]
labels = ['20-30岁', '31-40岁', '41-50岁', '51-60岁', '60岁以上']
df['Age_Group'] = pd.cut(df['Customer_Age'], bins=bins, labels=labels)

# 2. 创建“客户活跃度分级”
# 假设：过去12个月内的不活跃月数越少，客户越活跃
def activity_level(inactive_months):
    if inactive_months <= 1:
        return '高活跃度'
    elif inactive_months <= 2:
        return '中活跃度'
    else:
        return '低活跃度'
df['Activity_Level'] = df['Months_Inactive_12_mon'].apply(activity_level)

# 3. 创建“信用卡使用强度”
# 使用“Avg_Utilization_Ratio”（平均使用率）来分级
def usage_intensity(ratio):
    if ratio == 0:
        return '未使用'
    elif ratio <= 0.3:
        return '轻度使用'
    elif ratio <= 0.7:
        return '中度使用'
    else:
        return '重度使用'
df['Usage_Intensity'] = df['Avg_Utilization_Ratio'].apply(usage_intensity)

print("✅ 特征工程完成！")
print("\n✨ 新增的特征预览：")
display(df[['Customer_Age', 'Age_Group', 'Months_Inactive_12_mon', 'Activity_Level', 'Avg_Utilization_Ratio', 'Usage_Intensity']].head(10))
```
